### Sale Order

```python
# draw boxplot and distribution plot for each numeric column in sale_order.csv
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sale_order_df = pd.read_csv('export_csv/sale_order.csv')    
numeric_cols = sale_order_df.select_dtypes(include=['float64', 'int64']).columns
numeric_cols = [col for col in numeric_cols if col != 'channel_id' and col != 'merchant_id' and col != 'status']  # exclude id columns
for col in numeric_cols:
    plt.figure(figsize=(12, 5))
    
    plt.subplot(1, 1, 1)
    sns.boxplot(x=sale_order_df[col])
    plt.title(f'Boxplot of {col}')
    
    plt.tight_layout()
    plt.show()
```

### Product

```python
# draw boxplot and distribution plot for each numeric column in product.csv
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

product_df = pd.read_csv('export_csv/product.csv')
numeric_cols = product_df.select_dtypes(include=['float64', 'int64']).columns.difference([
    'product_id',
    'brand_id',
    'cat1_id',
    'cat2_id',
    'category_id',
    'product_sku'
])


for col in numeric_cols:
    plt.figure(figsize=(12, 5))
    
    plt.subplot(1, 1, 1)
    sns.boxplot(x=product_df[col])
    plt.title(f'Boxplot of {col}')
    
    plt.tight_layout()
    plt.show()
```

## 6. Bar chart

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import math

product_df = pd.read_csv("export_csv/product.csv")
product_category_df = pd.read_csv("export_csv/product_category.csv")

merged_df = product_df.merge(
    product_category_df[['category_id', 'category_name']],
    on='category_id',
    how='left'
)

category_counts = (
    merged_df.groupby(['category_id', 'category_name'])
    .size()
    .reset_index(name='product_count')
    .sort_values(by='category_id')
)

category_counts['category_name'] = category_counts['category_name'].fillna('Unknown')
category_counts['label'] = (
    category_counts['category_id'].astype(str) + ' - ' + category_counts['category_name']
)

n = len(category_counts)
chunk_size = math.ceil(n / 3)

chunks = [
    category_counts.iloc[i:i + chunk_size]
    for i in range(0, n, chunk_size)
]

fig, axes = plt.subplots(3, 1, figsize=(16, 18))

for i, chunk in enumerate(chunks):
    sns.barplot(data=chunk, x='label', y='product_count', ax=axes[i])
    axes[i].set_title(f'So luong san pham theo category - Phan {i+1}')
    axes[i].set_xlabel('Category')
    axes[i].set_ylabel('So luong san pham')
    axes[i].tick_params(axis='x', labelrotation=45)

plt.tight_layout()
plt.show()
```

```python
# vẽ biểu đồ cột phân phối số lượng đơn hàng theo từng status trong sale_order.csv
import pandas as pd
import matplotlib.pyplot as plt

sale_order_df = pd.read_csv('export_csv/sale_order.csv')
status_counts = sale_order_df['status'].value_counts()

plt.figure(figsize=(8, 5))
sns.barplot(x=status_counts.index, y=status_counts.values, palette='viridis')
plt.xlabel('Status')
plt.ylabel('Number of Orders')
plt.title('Number of Orders by Status')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# ve biểu đồ cột cho số lượng đơn hàng theo từng giới tính trong customer.csv
import pandas as pd
import matplotlib.pyplot as plt

customer_df = pd.read_csv('export_csv/customer.csv')
gender_counts = customer_df['gender'].value_counts()    

plt.figure(figsize=(8, 5))
sns.barplot(x=gender_counts.index, y=gender_counts.values, palette='viridis')
plt.xlabel('Gender')
plt.ylabel('Number of Customers')   
plt.title('Number of Customers by Gender')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
