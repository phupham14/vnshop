```python
# Load data
import pandas as pd
import numpy as np

sale_order_df = pd.read_csv('export_csv/sale_order.csv')
sale_order_line_df = pd.read_csv('export_csv/sale_order_line.csv')
product_df = pd.read_csv('export_csv/product.csv')
product_category_df = pd.read_csv('export_csv/product_category.csv')
customer_df = pd.read_csv('export_csv/customer.csv')

# Time features
sale_order_df['order_date'] = pd.to_datetime(sale_order_df['order_date'])
sale_order_df['order_hour'] = sale_order_df['order_date'].dt.hour
sale_order_df['day_of_week'] = sale_order_df['order_date'].dt.day_name()

# Merge master table
master_interaction = sale_order_line_df.merge(
    sale_order_df[['sale_order_id', 'user_id', 'order_date', 'order_hour', 'day_of_week']],
    on='sale_order_id'
).merge(
    product_df[['product_id', 'category_id']], on='product_id'
).merge(
    product_category_df[['category_id', 'category_name']], on='category_id'
).merge(
    customer_df[['user_id', 'age', 'gender']], on='user_id'
)

master_interaction.head()
```

#### So sánh dầu ăn với các danh mục liên quan khác

```python
target_cat = "Dầu ăn"

# Function to get category pairs and their counts
def get_category_pairs(df, category_col):
    from itertools import combinations
    pair_counts = {}
    
    for _, group in df.groupby('sale_order_id'):
        categories = group[category_col].unique()
        for cat_a, cat_b in combinations(categories, 2):
            pair = tuple(sorted((cat_a, cat_b)))
            pair_counts[pair] = pair_counts.get(pair, 0) + 1
            
    return pd.DataFrame([
        {'category_a': a, 'category_b': b, 'pair_count': count}
        for (a, b), count in pair_counts.items()
    ])

# Get category pairs and their counts
pair_stats = get_category_pairs(master_interaction, 'category_name')

if 'pair_stats' not in globals():
    raise NameError("pair_stats chưa được tạo. Hãy chạy cell tạo pair_stats trước.")

# Find related categories
mask_a = pair_stats['category_a'].eq(target_cat)
mask = mask_a | pair_stats['category_b'].eq(target_cat)

related_pairs = pair_stats.loc[mask, ['category_a', 'category_b', 'pair_count']].copy()
related_pairs['related_category'] = np.where(
    related_pairs['category_a'].eq(target_cat),
    related_pairs['category_b'],
    related_pairs['category_a']
)

display(
    related_pairs.groupby('related_category', as_index=False)['pair_count']
    .sum()
    .sort_values('pair_count', ascending=False)
    .head(15)
)
```

#### Lift cho các cặp category

```python
import pandas as pd
from itertools import combinations

# 1) Basket-level categories: mỗi order chỉ giữ unique category
basket_df = (
    master_interaction[['sale_order_id', 'category_name']]
    .dropna()
    .drop_duplicates()
)

n_orders = basket_df['sale_order_id'].nunique()

# 2) Support của từng category theo số order
cat_support = (
    basket_df.groupby('category_name')['sale_order_id']
    .nunique()
    .reset_index(name='order_count')
)
cat_support['support'] = cat_support['order_count'] / n_orders

# 3) Tạo category pairs trong cùng order
pair_rows = []

for _, group in basket_df.groupby('sale_order_id'):
    cats = sorted(group['category_name'].unique())
    if len(cats) >= 2:
        pair_rows.extend(combinations(cats, 2))

pair_df = pd.DataFrame(pair_rows, columns=['category_a', 'category_b'])

# 4) Pair stats
pair_stats = (
    pair_df.value_counts()
    .reset_index(name='pair_count')
)

pair_stats['pair_support'] = pair_stats['pair_count'] / n_orders

# 5) Join support từng category để tính lift
support_map = cat_support.set_index('category_name')['support']

pair_stats['support_a'] = pair_stats['category_a'].map(support_map)
pair_stats['support_b'] = pair_stats['category_b'].map(support_map)

pair_stats['lift'] = pair_stats['pair_support'] / (
    pair_stats['support_a'] * pair_stats['support_b']
)

# 6) Lọc các cặp quá hiếm để lift đỡ nhiễu
pair_lift = (
    pair_stats[pair_stats['pair_count'] >= 30]
    .sort_values(['lift', 'pair_count'], ascending=[False, False])
    .reset_index(drop=True)
)

display(pair_lift.head(20))
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# 1) Lấy category unique theo từng order của mỗi user
seq_df = (
    master_interaction[['user_id', 'order_date', 'sale_order_id', 'category_name']]
    .dropna()
    .drop_duplicates()
    .sort_values(['user_id', 'order_date', 'sale_order_id'])
)

transition_rows = []

for user_id, group in seq_df.groupby('user_id'):
    orders = (
        group.groupby(['order_date', 'sale_order_id'])['category_name']
        .unique()
        .reset_index()
        .sort_values(['order_date', 'sale_order_id'])
    )

    order_cats = orders['category_name'].tolist()

    for i in range(len(order_cats) - 1):
        current_cats = sorted(set(order_cats[i]))
        next_cats = sorted(set(order_cats[i + 1]))

        for a in current_cats:
            for b in next_cats:
                transition_rows.append((a, b))

transition_df = pd.DataFrame(transition_rows, columns=['from_category', 'to_category'])

transition_stats = (
    transition_df.value_counts()
    .reset_index(name='transition_count')
    .sort_values('transition_count', ascending=False)
)

# 2) Transition probability
transition_stats['transition_prob'] = (
    transition_stats['transition_count'] /
    transition_stats.groupby('from_category')['transition_count'].transform('sum')
)

display(transition_stats.head(20))
```

```python
top_k = 15

top_from = transition_stats.groupby('from_category')['transition_count'].sum().nlargest(top_k).index
top_to = transition_stats.groupby('to_category')['transition_count'].sum().nlargest(top_k).index

heatmap_df = transition_stats[
    transition_stats['from_category'].isin(top_from) &
    transition_stats['to_category'].isin(top_to)
].copy()

transition_matrix = heatmap_df.pivot(
    index='from_category',
    columns='to_category',
    values='transition_prob'
).fillna(0)

plt.figure(figsize=(14, 10))
sns.heatmap(
    transition_matrix,
    cmap='YlGnBu',
    annot=True,
    fmt='.2f'
)
plt.title('Transition Probability Heatmap: From Category -> Next Order Category')
plt.xlabel('To Category')
plt.ylabel('From Category')
plt.xticks(rotation=45, ha='right')
plt.yticks(rotation=0)
plt.tight_layout()
plt.show()
```

## 7. Assumption

### Shipping Province
- Hypothesis: Người dùng trong cùng shipping_province có phân phối danh mục mua hàng tương đồng
- Why: Do khí hậu + availability + văn hóa tiêu dùng
- Validate:
    - So sánh distribution category_name theo shipping_province_name
    - Dùng cosine similarity giữa các tỉnh
Impact:
    - Feature: shipping_province_name
Nhận xét:
- Việc phân tích dữ liệu cho thấy giả thuyết "Người dùng trong cùng tỉnh thành có phân phối danh mục mua hàng tương đồng" là hoàn toàn có cơ sở và mang tính thực tế cao.
- Các cụm tương đồng: * Bắc Ninh - Nam Định - Thái Bình: Có sự tương đồng rất sâu sắc (màu xanh đậm), phản ánh thói quen tiêu dùng khá đồng nhất của các tỉnh khu vực phía Bắc.
- TP.HCM và Hà Nội: Dù là hai đầu cầu đất nước, sự tương đồng vẫn rất cao, cho thấy lối sống đô thị hóa (urban lifestyle) dẫn đến nhu cầu hàng hóa tương tự nhau.
- Biểu đồ cột (Bar chart) cho thấy rõ lý do tại sao độ tương đồng lại cao như vậy:

    - Thống trị bởi nhu yếu phẩm: Danh mục "Dầu ăn" và "Giặt giũ" chiếm tỷ trọng áp đảo ở tất cả các tỉnh. Đây là các mặt hàng thiết yếu, ít bị ảnh hưởng bởi biến động vùng miền.

    - Sự khác biệt tinh tế: * Nhóm mẹ và bé: Các danh mục như Sữa bột cho bé, Sữa bột, Nước rửa bình cho thấy sự ổn định về nhu cầu ở cả 5 tỉnh thành trọng điểm.

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.metrics.pairwise import cosine_similarity

# ======================
# 1. Chuẩn bị dimension tables (set index để join nhanh)
# ======================
product_map = product_df.set_index('product_id')['category_id']
category_map = product_category_df.set_index('category_id')['category_name']
order_map = sale_order_df.set_index('sale_order_id')['shipping_province_name']

# ======================
# 2. Map thay vì merge (NHANH hơn nhiều)
# ======================
df = sale_order_line_df.copy()

df['category_id'] = df['product_id'].map(product_map)
df['category_name'] = df['category_id'].map(category_map)
df['shipping_province_name'] = df['sale_order_id'].map(order_map)

# Drop NA sớm để giảm size
df = df.dropna(subset=['category_name', 'shipping_province_name'])

# ======================
# 3. Lấy Top provinces (giảm data sớm)
# ======================
top_provinces = (
    df['shipping_province_name']
    .value_counts()
    .head(5)
    .index
)

df = df[df['shipping_province_name'].isin(top_provinces)]

# ======================
# 4. Groupby thay vì pivot_table (nhanh hơn)
# ======================
province_dist = (
    df.groupby(['shipping_province_name', 'category_name'])['sale_order_id']
    .count()
    .unstack(fill_value=0)
)

# ======================
# 5. Normalize (vectorized, OK)
# ======================
province_dist_pct = province_dist.div(province_dist.sum(axis=1), axis=0)

# ======================
# 6. Top categories (giảm dimension)
# ======================
top_categories = (
    province_dist_pct.sum(axis=0)
    .nlargest(10)
    .index
)

filtered_plot_df = province_dist_pct[top_categories]

# ======================
# 7. Cosine similarity
# ======================
similarity_matrix = cosine_similarity(province_dist_pct.values)

similarity_df = pd.DataFrame(
    similarity_matrix,
    index=province_dist_pct.index,
    columns=province_dist_pct.index
)

# ======================
# 8. Visualization
# ======================
plt.figure(figsize=(12, 10))
sns.heatmap(
    similarity_df,
    cmap='YlGnBu',
    annot=True,          # hiển thị giá trị
    fmt=".2f",           # làm tròn 2 chữ số thập phân
    linewidths=0.5,      # cho dễ nhìn
    linecolor='white'
)
plt.title('Độ tương đồng về danh mục mua sắm giữa các tỉnh thành')
plt.show()
filtered_plot_df.T.plot(kind='bar', figsize=(15, 6))
plt.title('So sánh cơ cấu Top 10 danh mục (%) giữa các tỉnh trọng điểm')
plt.ylabel('Tỷ lệ mua sắm')
plt.xticks(rotation=45)
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```

    - Marketing segmentation thường dựa trên demographic

Nhận xét:
Giả thuyết "Người dùng cùng nhóm tuổi và giới tính có phân phối mua hàng tương đồng (khác biệt so với quần thể chung)" được xác nhận mạnh mẽ thông qua chỉ số Jensen-Shannon (JS) Divergence.

- Chỉ số JS Divergence cao: Biểu đồ cho thấy có những nhóm nhân khẩu học đạt mức divergence lên tới 0.6 - 0.7. Trong thống kê, JS Divergence càng cao nghĩa là phân phối của nhóm đó càng khác biệt xa so với phân phối tổng (Global Distribution).

- Minh chứng từ dữ liệu: Các nhóm nam giới lớn tuổi (ví dụ: Group Index 83, 79, 81 tương ứng với độ tuổi 57-59) có sự khác biệt cực kỳ lớn so với mức trung bình của toàn bộ sàn. Điều này chứng tỏ họ không mua sắm những thứ mà "đám đông" thường mua (như tã sữa hay dầu ăn với số lượng lớn), mà tập trung vào các danh mục đặc thù.

- Khác với sự tương đồng (Similarity) ở giả thuyết địa lý, JS Divergence đo lường "cá tính" của phân khúc:

    - Nhóm lệch cao (High Divergence): Thường là các nhóm có hành vi mua sắm rất tập trung (niche). Ví dụ: Nam giới 59 tuổi có thể chỉ tập trung vào đồ điện tử, dụng cụ sửa chữa hoặc thực phẩm chức năng, thay vì giỏ hàng đa dạng nhu yếu phẩm như nhóm hộ gia đình.

    - Nhóm lệch thấp (Low Divergence): Các nhóm ở phía bên phải biểu đồ (JS < 0.1) có hành vi mua sắm "hòa tan" vào đám đông. Đây thường là nhóm khách hàng đại trà (Mass Market), mua sắm đủ mọi danh mục thiết yếu.

- Nguyên nhân xác nhận giả thuyết:
    - Sự ưu tiên ngành hàng (Gender Bias): Nam giới và nữ giới có xu hướng chi tiêu vào các ngành hàng khác biệt rõ rệt (ví dụ: Mỹ phẩm vs. Phụ kiện xe).

    - Giai đoạn cuộc đời (Life Stages): Nhóm tuổi phản ánh nhu cầu thực tế. Gen Z (thích bắt trend, thời trang) sẽ có phân phối hàng hóa khác hoàn toàn so với Gen X (tập trung cho gia đình, sức khỏe).

```python
import numpy as np
from scipy.spatial.distance import jensenshannon

product_category_df = pd.read_csv("export_csv/product_category.csv")
sale_order_df = pd.read_csv("export_csv/sale_order.csv")
sale_order_line_df = pd.read_csv("export_csv/sale_order_line.csv")
product_df = pd.read_csv("export_csv/product.csv")
customer_df = pd.read_csv("export_csv/customer.csv")

merged_df = sale_order_line_df.merge(
    sale_order_df[['sale_order_id', 'user_id']],
    on='sale_order_id',
    how='left'
).merge(
    customer_df[['user_id', 'age', 'gender']],
    on='user_id',
    how='left'
).merge(
    product_df[['product_id', 'category_id']],
    on='product_id',
    how='left'
).merge(
    product_category_df[['category_id', 'category_name']],
    on='category_id',
    how='left'
)

# ======================
# 1. GLOBAL DISTRIBUTION
# ======================
global_dist = (
    merged_df['category_name']
    .value_counts(normalize=True)
    .sort_index()
)

# ======================
# 2. DEMOGRAPHIC DISTRIBUTION
# ======================
group_dist = (
    merged_df.groupby(['age', 'gender'])['category_name']
    .value_counts(normalize=True)
    .unstack(fill_value=0)
)

# align columns
group_dist = group_dist.reindex(columns=global_dist.index, fill_value=0)

# ======================
# 3. COMPUTE DIVERGENCE
# ======================
results = []

for idx, row in group_dist.iterrows():
    js_div = jensenshannon(row.values, global_dist.values)
    
    results.append({
        'age_group': idx[0],
        'gender': idx[1],
        'js_divergence': js_div
    })

div_df = pd.DataFrame(results)

# ======================
# 4. SORT & VIEW
# ======================
div_df = div_df.sort_values(by='js_divergence', ascending=False)

print(div_df.head())
```

```python
# nhóm khách hàng từ 57-59 tuổi có sự khác biệt lớn nhất về cơ cấu danh mục so với trung bình, đặc biệt là nhóm nữ. Đây có thể là nhóm khách hàng lớn tuổi, có nhu cầu và sở thích mua sắm khác biệt so với phần còn lại. Các nhà bán lẻ nên xem xét các chiến lược tiếp cận riêng cho nhóm này, như cung cấp các sản phẩm phù hợp với nhu cầu của người cao tuổi hoặc tạo ra các chương trình khuyến mãi đặc biệt để thu hút họ.
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(12, 6))
sns.barplot(x='age_group', y='js_divergence', hue='gender', data=div_df)
plt.title("Divergence from Global Distribution by Age Group and Gender")
plt.xlabel("Age Group")
plt.ylabel("JS Divergence")
plt.legend(title='Gender')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

### Giả định về thương hiệu (Brand)
Người dùng có xu hướng duy trì sự ưa thích đối với một hoặc một nhóm thương hiệu nhất định.

Thương hiệu phản ánh các yếu tố như chất lượng sản phẩm, độ tin cậy và mức độ quen thuộc của người dùng. Do đó, người dùng thường có xu hướng tiếp tục mua các sản phẩm thuộc thương hiệu mà họ đã từng sử dụng hoặc các thương hiệu cùng phân khúc (ví dụ: cao cấp hoặc bình dân).

```python
sns.set_theme(style="whitegrid")
pd.set_option("display.max_columns", None)
pd.set_option("display.float_format", lambda x: f"{x:,.4f}")

brand = pd.read_csv("export_csv/brand.csv")
product = pd.read_csv("export_csv/product.csv")
sale_order_line = pd.read_csv("export_csv/sale_order_line.csv")
sale_order = pd.read_csv("export_csv/sale_order.csv", parse_dates=["order_date"])
```

```python
# Join data for brand loyalty analysis
# Loai bo gift line vi gift khong phan anh lua chon thuong hieu chu dong cua user
sale_order_line = sale_order_line[
    sale_order_line["is_gift"].astype(str).str.lower() != "true"
].copy()

df = (
    sale_order_line
    .merge(
        sale_order[["sale_order_id", "user_id", "order_date", "status"]],
        on="sale_order_id",
        how="left"
    )
    .merge(
        product[["product_id", "brand_id", "product_amount"]],
        on="product_id",
        how="left"
    )
    .merge(
        brand,
        on="brand_id",
        how="left"
    )
    .dropna(subset=["user_id", "brand_id"])
    .copy()
)

df["brand_id"] = df["brand_id"].astype("Int64")

print(df.shape)
display(df.head())
```

```python
# Basic sanity check
print("So user:", df["user_id"].nunique())
print("So brand:", df["brand_id"].nunique())
print("So dong mua hang:", len(df))
print("Top 10 brand theo so luong mua:")
display(
    df["brand_name"]
    .value_counts()
    .head(10)
    .rename_axis("brand_name")
    .reset_index(name="purchase_events")
)
```

```python
# Tao chuoi mua hang theo thu tu thoi gian cho tung user
events = (
    df.sort_values(["user_id", "order_date", "sale_order_id", "line_id"])[
        ["user_id", "order_date", "sale_order_id", "line_id", "brand_id", "brand_name"]
    ]
    .copy()
)

events["prev_brand_id"] = events.groupby("user_id")["brand_id"].shift(1)
events["prev_brand_name"] = events.groupby("user_id")["brand_name"].shift(1)

# tranition analysis: xem xac suat mua lai cung 1 brand
transitions = events.dropna(subset=["prev_brand_id"]).copy()
transitions["same_brand"] = (transitions["brand_id"] == transitions["prev_brand_id"]).astype(int)

print("Tong so transitions:", len(transitions))
display(transitions.head())
```

```python
observed_repeat_rate = transitions["same_brand"].mean()
print(f"Observed same-brand repeat rate: {observed_repeat_rate:.4%}")
```

```python
# Tinh baseline ngau nhien toan he thong
# Neu user mua ngau nhien theo phan phoi brand cua toan he thong,
# xac suat mua lai dung brand truoc do se bang market share cua brand do.
global_brand_share = df["brand_id"].value_counts(normalize=True)
transitions["global_prev_brand_share"] = transitions["prev_brand_id"].map(global_brand_share)

random_baseline = transitions["global_prev_brand_share"].mean()
print(f"Random baseline: {random_baseline:.4%}")
print(f"Lift vs random baseline: {observed_repeat_rate / random_baseline:.2f}x")
```

```python
from scipy.stats import norm

# Kiểm định z-test 1 phía
# H0: observed_repeat_rate <= random_baseline
# H1: observed_repeat_rate  > random_baseline
n = len(transitions) # số lượng transitions
se = np.sqrt(random_baseline * (1 - random_baseline) / n) # standard error cho tỉ lệ
z_score = (observed_repeat_rate - random_baseline) / se # z-score cho tỉ lệ
p_value = 1 - norm.cdf(z_score) # p-value cho z-score (1-sided)

result = pd.DataFrame({
    "metric": [
        "observed_same_brand_rate",
        "random_baseline",
        "lift_vs_random",
        "z_score",
        "p_value_one_sided"
    ],
    "value": [
        observed_repeat_rate,
        random_baseline,
        observed_repeat_rate / random_baseline,
        z_score,
        p_value
    ]
})

display(result)
```

```python
# Cell 10: Dien giai ket qua
if p_value < 0.05 and observed_repeat_rate > random_baseline:
    print(
        f"Ket luan: Chap nhan gia thuyet nghien cuu. "
        f"Ti le mua lai cung brand o lan mua ke tiep la {observed_repeat_rate:.2%}, "
        f"cao hon ro ret so voi baseline ngau nhien {random_baseline:.2%} "
        f"(lift = {observed_repeat_rate / random_baseline:.2f}x, p-value = {p_value:.4g})."
    )
else:
    print(
        f"Ket luan: Chua du bang chung ung ho gia thuyet. "
        f"Observed = {observed_repeat_rate:.2%}, baseline = {random_baseline:.2%}, p-value = {p_value:.4g}."
    )
```

```python
# Brand repeat rate theo tung brand
brand_repeat = (
    transitions.groupby(["prev_brand_id", "prev_brand_name"])
    .agg(
        transitions=("same_brand", "size"),
        repeat_rate=("same_brand", "mean")
    )
    .reset_index()
)

# Loc nhung brand co du transitions de ket luan co y nghia
brand_repeat_filtered = brand_repeat[brand_repeat["transitions"] >= 100].copy()
brand_repeat_filtered = brand_repeat_filtered.sort_values(
    ["repeat_rate", "transitions"], ascending=[False, False]
)

display(brand_repeat_filtered.head(15))

top_n = 15
plot_df = brand_repeat_filtered.head(top_n).sort_values("repeat_rate", ascending=True)

plt.figure(figsize=(10, 7))
sns.barplot(data=plot_df, x="repeat_rate", y="prev_brand_name", palette="Blues_r")
plt.title("Top 15 Brands with Highest Repeat Rate")
plt.xlabel("Repeat Rate")
plt.ylabel("Brand")
plt.xlim(0, plot_df["repeat_rate"].max() * 1.1)
plt.show()
```

```python
# Do tap trung vao brand ua thich nhat cua moi user
user_brand_counts = (
    df.groupby(["user_id", "brand_id"])
    .size()
    .rename("purchase_cnt")
    .reset_index()
)
# Tinh tong so mua hang cua moi user de tinh share tung brand
user_total = (
    user_brand_counts.groupby("user_id")["purchase_cnt"]
    .sum()
    .rename("total_purchase_cnt")
    .reset_index()
)

# Merge de tinh share tung brand trong tong so mua hang cua user
user_brand_counts = user_brand_counts.merge(user_total, on="user_id", how="left")
user_brand_counts["brand_share"] = (
    user_brand_counts["purchase_cnt"] / user_brand_counts["total_purchase_cnt"]
)

# Tinh do tap trung cua user vao 1 hay 1 so it brand
user_concentration = (
    user_brand_counts.groupby("user_id")
    .agg(
        total_purchase_cnt=("purchase_cnt", "sum"),
        n_brands=("brand_id", "nunique"),
        top_brand_share=("brand_share", "max"),
        hhi=("brand_share", lambda s: np.square(s).sum())
    )
    .reset_index()
)

# Lọc user có ít nhất 3 lần mua hàng để tránh nhiễu từ user mua ít
user_concentration = user_concentration[user_concentration["total_purchase_cnt"] >= 3].copy()
user_concentration["uniform_top_share"] = 1 / user_concentration["n_brands"]

display(user_concentration.describe())
```

```python
# So sanh top brand share thuc te voi muc deu
summary_concentration = pd.DataFrame({
    "metric": [
        "avg_top_brand_share",
        "avg_uniform_top_share",
        "median_hhi",
        "share_users_top_brand_gt_50pct"
    ],
    "value": [
        user_concentration["top_brand_share"].mean(),
        user_concentration["uniform_top_share"].mean(),
        user_concentration["hhi"].median(),
        (user_concentration["top_brand_share"] > 0.5).mean()
    ]
})

display(summary_concentration)

# Histogram top brand share
plt.figure(figsize=(9, 5))
sns.histplot(user_concentration["top_brand_share"], bins=30, color="#1f77b4")
plt.axvline(user_concentration["top_brand_share"].mean(), color="red", linestyle="--", label="Mean")
plt.title("Distribution of Top Brand Share per User (users with >= 3 purchases)")
plt.xlabel("Top Brand Share")
plt.ylabel("Number of Users")
plt.legend()
plt.show()
    
```

Kiểm định thống kê với z-score rất lớn (359) và p-value gần bằng 0 cho thấy hiện tượng này có ý nghĩa thống kê mạnh và không phải do ngẫu nhiên.

### Giả định về độ nhạy khuyến mãi:
Người dùng có mức độ nhạy cảm khác nhau đối với các chương trình khuyến mãi như giảm giá đơn hàng hoặc miễn/giảm phí vận chuyển.

Một số người dùng có xu hướng nhạy cảm với giá và thường mua hàng khi có khuyến mãi, trong khi những người khác ưu tiên thương hiệu hoặc sự tiện lợi hơn là yếu tố giá cả.

```python
# Promotion sensitivity by user based on order discount, shipping discount, and gift items
import pandas as pd

if 'sale_order_df' not in globals():
    sale_order_df = pd.read_csv('export_csv/sale_order.csv')

if 'sale_order_line_df' not in globals():
    sale_order_line_df = pd.read_csv('export_csv/sale_order_line.csv')

order_promo = sale_order_df[[
    'sale_order_id',
    'user_id',
    'order_discount',
    'shipping_discount',
    'order_amount'
]].copy()

# Tính xem đơn hàng nào có quà tặng (is_gift = True) trong sale_order_line_df
gift_by_order = sale_order_line_df[['sale_order_id', 'is_gift']].copy()
gift_by_order['is_gift'] = gift_by_order['is_gift'].astype(str).str.lower().eq('true')
gift_by_order = gift_by_order.groupby('sale_order_id', as_index=False)['is_gift'].max()

# Merge thông tin quà tặng vào order_promo
order_promo = order_promo.merge(gift_by_order, on='sale_order_id', how='left')
order_promo['is_gift'] = order_promo['is_gift'].fillna(False)
order_promo['has_order_discount'] = order_promo['order_discount'] > 0
order_promo['has_shipping_discount'] = order_promo['shipping_discount'] > 0
order_promo['has_any_promo'] = (
    order_promo['has_order_discount']
    | order_promo['has_shipping_discount']
    | order_promo['is_gift']
)

# Tính baseline cho từng loại khuyến mãi
promotion_baseline = pd.DataFrame({
    'metric': ['order_discount_rate', 'shipping_discount_rate', 'gift_rate', 'any_promo_rate'],
    'value': [
        order_promo['has_order_discount'].mean(),
        order_promo['has_shipping_discount'].mean(),
        order_promo['is_gift'].mean(),
        order_promo['has_any_promo'].mean()
    ]
})

# Hàm phân loại độ nhạy cảm với khuyến mãi cho từng user dựa trên tỷ lệ sử dụng các loại khuyến mãi
def classify_promotion_sensitivity(row, min_orders=3):
    if row['total_orders'] < min_orders:
        return 'insufficient_data' # dữ liệu không đủ để phân loại

    order_rate = row['order_discount_rate']
    shipping_rate = row['shipping_discount_rate']
    gift_rate = row['gift_rate']
    any_promo_rate = row['any_promo_rate']

    if any_promo_rate >= 0.95:
        if order_rate >= 0.8 and order_rate >= shipping_rate:
            return 'highly_price_sensitive' # ưu tiên giảm giá đơn hàng
        if shipping_rate >= 0.8 and shipping_rate > order_rate:
            return 'highly_shipping_sensitive' # ưu tiên giảm giá vận chuyển
        if gift_rate >= 0.2:
            return 'highly_gift_sensitive' # ưu tiên quà tặng
        return 'highly_promotion_sensitive' # nhạy cảm với khuyến mãi

    if order_rate >= 0.8 and order_rate >= shipping_rate:
        return 'price_sensitive' # nhạy cảm với giảm giá đơn hàng

    if shipping_rate >= 0.8 and shipping_rate > order_rate:
        return 'shipping_sensitive' # nhạy cảm với giảm giá vận chuyển

    if gift_rate >= 0.2:
        return 'gift_sensitive' # nhạy cảm với quà tặng

    if any_promo_rate <= 0.5:
        return 'low_promotion_sensitive' # ít nhạy cảm với khuyến mãi

    return 'mixed_promotion_response' # phản ứng hỗn hợp, không rõ ưu tiên loại khuyến mãi nào

# Tính các chỉ số nhạy cảm với khuyến mãi cho từng user
user_promotion_sensitivity = (
    order_promo.groupby('user_id')
    .agg(
        total_orders=('sale_order_id', 'count'),
        order_discount_rate=('has_order_discount', 'mean'),
        shipping_discount_rate=('has_shipping_discount', 'mean'),
        gift_rate=('is_gift', 'mean'),
        any_promo_rate=('has_any_promo', 'mean'),
        avg_order_amount=('order_amount', 'mean')
    )
    .reset_index()
)

# Phân loại người dùng theo độ nhạy cảm với khuyến mãi
user_promotion_sensitivity['promotion_sensitivity_segment'] = user_promotion_sensitivity.apply(
    classify_promotion_sensitivity,
    axis=1
)

# Tóm tắt số lượng và tỷ lệ người dùng theo từng phân khúc nhạy cảm với khuyến mãi
promotion_segment_summary = (
    user_promotion_sensitivity['promotion_sensitivity_segment']
    .value_counts(dropna=False)
    .rename_axis('promotion_sensitivity_segment')
    .reset_index(name='user_count')
)
promotion_segment_summary['user_share'] = (
    promotion_segment_summary['user_count'] / promotion_segment_summary['user_count'].sum()
)

display(promotion_baseline)
display(promotion_segment_summary)
display(user_promotion_sensitivity.head())
```

```python
import numpy as np
import pandas as pd

def add_age_group(df):
    df = df.copy()
    
    bins = [0, 18, 25, 35, 45, 60, 100]
    labels = ['<18', '18-25', '26-35', '36-45', '46-60', '>60']
    
    df['age_group'] = pd.cut(
        df['age'],
        bins=bins,
        labels=labels,
        right=True,          # (a, b]
        include_lowest=True  # include 0
    )
    
    return df
```

```python
# Hàm xây dựng mapping từ brand_id sang phân khúc dựa trên giá trung vị của brand
def build_brand_segment_map(train_df):
    required_cols = {'brand_id', 'product_amount'}
    if not required_cols.issubset(train_df.columns):
        return {}

    # Tính giá trung vị của mỗi brand dựa trên các dòng có giá trị không null
    brand_price = (
        train_df.dropna(subset=['brand_id', 'product_amount'])
        .groupby('brand_id')['product_amount']
        .median()
        .reset_index(name='brand_median_price')
    )

    if brand_price.empty:
        return {}

    # Phân loại brand thành 3 phân khúc dựa trên giá trung vị
    labels = ['budget', 'mid', 'premium']
    n_brands = len(brand_price)

    if n_brands >= 3:
        ranked_price = brand_price['brand_median_price'].rank(method='first')
        brand_price['brand_segment'] = pd.qcut(ranked_price, q=3, labels=labels)
    elif n_brands == 2:
        brand_price = brand_price.sort_values('brand_median_price').copy()
        brand_price['brand_segment'] = ['budget', 'premium']
    else:
        brand_price['brand_segment'] = 'mid'

    return brand_price.set_index('brand_id')['brand_segment'].astype(str).to_dict()

# Hàm áp dụng phân khúc brand vào dataframe
def apply_brand_segment(df, brand_segment_map):
    df = df.copy()
    if 'brand_id' not in df.columns:
        df['brand_segment'] = 'unknown'
        return df

    df['brand_segment'] = (
        df['brand_id'].map(brand_segment_map).fillna('unknown').astype(str)
    )
    return df
```

```python
def temporal_split(df):
    df = df.sort_values('order_date')
    
    test_df = df.groupby('user_id').tail(1)
    train_df = df.drop(test_df.index)
    
    return train_df, test_df
```

```python
# Tính lift và build pair dict
def build_pair_dict(pair_stats, min_count=50):
    df = pair_stats[pair_stats['pair_count'] >= min_count].copy()
    
    df['score'] = df['lift'] * np.log1p(df['pair_count'])
    
    pair_dict = {}
    
    for _, row in df.iterrows():
        a, b = row['category_a'], row['category_b']
        s = row['score']
        
        pair_dict.setdefault(a, {})[b] = s
        pair_dict.setdefault(b, {})[a] = s
        
    return pair_dict
```

```python
# Xây dựng phân phối có điều kiện với smoothing
def _build_context_distribution(train_df, global_dist, context_cols, alpha=20):
    train_df = train_df.copy()
    for col in context_cols:
        if col not in train_df.columns:
            train_df[col] = 'unknown'
        else:
            train_df[col] = train_df[col].fillna('unknown')

    # Tính phân phối có điều kiện P(category | context) với smoothing
    ctx = (
        train_df.groupby(context_cols + ['category_name'])
        .size()
        .reset_index(name='count')
    )

    # Tính tổng count cho mỗi context để chuẩn hóa
    ctx_total = ctx.groupby(context_cols)['count'].transform('sum') # tổng số lần mua hàng trong mỗi context
    ctx['global_prob'] = ctx['category_name'].map(global_dist) # xác suất toàn hệ thống của category
    ctx['prob'] = (ctx['count'] + alpha * ctx['global_prob']) / (ctx_total + alpha) # smoothing
    ctx['lift'] = np.log1p(ctx['prob'] / (ctx['global_prob'] + 1e-6)) # lift có log để tránh số quá lớn, +1e-6 để tránh chia cho 0

    return ctx

# Xây dựng phân phối có điều kiện cho nhiều context khác nhau
def build_distributions(train_df, alpha=20):
    global_dist = train_df['category_name'].value_counts(normalize=True) # phân phối toàn hệ thống P(category)

    demo_dist = _build_context_distribution(train_df, global_dist, ['gender', 'age_group'], alpha)
    province_dist = _build_context_distribution(train_df, global_dist, ['shipping_province_name'], alpha)
    dow_dist = _build_context_distribution(train_df, global_dist, ['day_of_week'], alpha)
    hour_dist = _build_context_distribution(train_df, global_dist, ['hour_bucket'], alpha)
    brand_segment_dist = _build_context_distribution(train_df, global_dist, ['brand_segment'], alpha)

    return global_dist, demo_dist, province_dist, dow_dist, hour_dist, brand_segment_dist
```

```python
# Xây dựng thống kê chuyển đổi giữa các category dựa trên thứ tự mua hàng của user
def build_transition_stats(train_df):
    order_level = (
        train_df.groupby(['user_id', 'sale_order_id', 'order_date'])['category_name']
        .apply(lambda x: sorted(set(x)))
        .reset_index(name='categories')
        .sort_values(['user_id', 'order_date', 'sale_order_id'])
    )

    # Tạo danh sách chuyển đổi giữa các category dựa trên thứ tự mua hàng của mỗi user
    transition_rows = []
    for _, user_orders in order_level.groupby('user_id'):
        categories_by_order = user_orders['categories'].tolist()
        for prev_cats, next_cats in zip(categories_by_order, categories_by_order[1:]):
            transition_rows.extend((prev_cat, next_cat) for prev_cat in prev_cats for next_cat in next_cats)

    if not transition_rows:
        return pd.DataFrame(columns=['from_category', 'to_category', 'transition_count', 'transition_prob'])

    # Tính thống kê chuyển đổi và xác suất chuyển đổi giữa các category
    transition_stats = (
        pd.DataFrame(transition_rows, columns=['from_category', 'to_category'])
        .value_counts()
        .reset_index(name='transition_count')
    )

    # Tính xác suất chuyển đổi P(to_category | from_category)
    transition_stats['transition_prob'] = (
        transition_stats['transition_count']
        / transition_stats.groupby('from_category')['transition_count'].transform('sum')
    )

    return transition_stats

# Xây dựng dictionary chuyển đổi giữa các category với điểm số dựa trên lift và xác suất chuyển đổi
def build_transition_dict(transition_stats):
    df = transition_stats.copy()
    if df.empty:
        return {}

    # Tính xác suất chuyển đổi toàn hệ thống cho mỗi category đích
    global_prob = (
        df.groupby('to_category')['transition_count']
        .sum() / df['transition_count'].sum()
    )

    # Tính lift và điểm số cho mỗi cặp chuyển đổi
    df['lift'] = df.apply(
        lambda x: np.log1p(x['transition_prob'] / (global_prob[x['to_category']] + 1e-6)),
        axis=1
    )

    # Kết hợp lift và xác suất chuyển đổi để tạo điểm số cuối cùng cho mỗi cặp chuyển đổi
    df['score'] = df['lift'] * df['transition_prob']

    trans_dict = {}

    # Chỉ giữ lại những cặp chuyển đổi có đủ số lượng để đảm bảo độ tin cậy
    df_filtered = df[df['transition_count'] >= 5]  # Ví dụ: chỉ giữ lại các chuyển đổi có ít nhất 5 lần xảy ra
    for _, row in df_filtered.iterrows():
        a, b = row['from_category'], row['to_category']
        s = row['score']
        trans_dict.setdefault(a, {})[b] = s

    return trans_dict
```

```python
# Hàm xây dựng preference vector cho mỗi user dựa trên lịch sử mua hàng của họ
def build_user_pref(train_df):
    user_pref = train_df.groupby(['user_id', 'category_name']).size()
    
    # normalize per user
    user_pref = user_pref / user_pref.groupby('user_id').sum()
    
    return user_pref.to_dict()
```

```python
# H?m x?y d?ng dictionary ?? lookup nhanh c?c context distribution
def build_context_dict(context_dist, key_cols):
    return context_dist.set_index(key_cols + ['category_name']).to_dict('index')

def build_segment_dict(segment_dist):
    return build_context_dict(segment_dist, ['gender', 'age_group'])
```

```python
# H?m x?y d?ng dictionary ?? lookup transition score nhanh ch?ng
def build_user_last_categories(train_df):
    last_orders = (
        train_df.sort_values(['user_id', 'order_date', 'sale_order_id'])
        .groupby('user_id')
        .tail(1)[['user_id', 'sale_order_id']]
        .drop_duplicates('user_id')
    )

    # Tạo mapping từ user_id sang sale_order_id của đơn hàng cuối cùng
    last_order_map = last_orders.set_index('user_id')['sale_order_id'].to_dict()

    # Lấy các dòng tương ứng với đơn hàng cuối cùng của mỗi user
    last_order_rows = train_df[
        train_df.apply(lambda row: last_order_map.get(row['user_id']) == row['sale_order_id'], axis=1)
    ]

    return (
        last_order_rows.groupby('user_id')['category_name']
        .apply(lambda x: list(dict.fromkeys(x.tolist())))
        .to_dict()
    )
```

```python
# Hàm tổng hợp tất cả các bước để xây dựng preference vector và transition dict
def build_pair_stats(train_df): 
    transition_stats = build_transition_stats(train_df)
    
    if transition_stats.empty:
        return pd.DataFrame(columns=['category_a', 'category_b', 'pair_count', 'lift'])
    
    global_prob = train_df['category_name'].value_counts(normalize=True)
    
    # Tính lift cho mỗi cặp chuyển đổi
    transition_stats['lift'] = transition_stats.apply(
        lambda x: np.log1p(x['transition_prob'] / (global_prob[x['to_category']] + 1e-6)),
        axis=1
    )
    
    transition_stats.rename(columns={'from_category': 'category_a', 'to_category': 'category_b', 'transition_count': 'pair_count'}, inplace=True)
    
    return transition_stats[['category_a', 'category_b', 'pair_count', 'lift']]
```

```python
# Hàm phân tích lift để hiểu rõ hơn về những segment nào đang có lift cao, lift thấp, và lift trung bình theo category
def analyze_lift(segment_dist, top_k=10):
    
    segment_dist = segment_dist.copy()
    print("Top lift (overall):")
    display(
        segment_dist.sort_values('lift', ascending=False)[
            ['gender', 'age_group', 'category_name', 'lift']
        ].head(top_k)
    )
    
    print("\nLowest lift:")
    display(
        segment_dist.sort_values('lift', ascending=True)[
            ['gender', 'age_group', 'category_name', 'lift']
        ].head(top_k)
    )
    
    print("\nAvg lift per category:")
    display(
        segment_dist.groupby('category_name')['lift']
        .mean()
        .sort_values(ascending=False)
        .head(top_k)
    )

analyze_lift(segment_dist)
```

```python
# H?m ?? xu?t category cho m?t user c? th? d?a tr?n c?c gi? ??nh ?? ki?m ch?ng
def recommend_v2(
    user_id,
    user_row,
    user_pref,
    segment_dict,
    global_dist,
    categories,
    pair_dict,
    trans_dict,
    user_last_cats,
    province_dict=None,
    dow_dict=None,
    hour_dict=None,
    brand_segment_dict=None,
    top_k=5
):
    gender = user_row.get('gender', 'Unknown')
    age_group = user_row.get('age_group', 'unknown')
    province = user_row.get('shipping_province_name', 'Unknown')
    day_of_week = user_row.get('day_of_week', 'Unknown')
    hour_bucket = user_row.get('hour_bucket', 'unknown')
    brand_segment = user_row.get('brand_segment', 'unknown')

    last_cats = user_last_cats.get(user_id, [])

    province_dict = province_dict or {}
    dow_dict = dow_dict or {}
    hour_dict = hour_dict or {}
    brand_segment_dict = brand_segment_dict or {}

    scores = {}

    for cat in categories:
        user_score = user_pref.get((user_id, cat), 0)

        demo_entry = segment_dict.get((gender, age_group, cat), {})
        province_entry = province_dict.get((province, cat), {})
        dow_entry = dow_dict.get((day_of_week, cat), {})
        hour_entry = hour_dict.get((hour_bucket, cat), {})
        brand_entry = brand_segment_dict.get((brand_segment, cat), {})

        demo_prob = demo_entry.get('prob', 0)
        demo_lift = demo_entry.get('lift', 0)
        province_prob = province_entry.get('prob', 0)
        province_lift = province_entry.get('lift', 0)
        dow_prob = dow_entry.get('prob', 0)
        hour_prob = hour_entry.get('prob', 0)
        brand_prob = brand_entry.get('prob', 0)
        brand_lift = brand_entry.get('lift', 0)
        global_score = global_dist.get(cat, 1e-6)

        pair_values = [pair_dict.get(lc, {}).get(cat, 0) for lc in last_cats if lc != cat]
        pair_score = float(np.mean(np.log1p(pair_values))) if pair_values else 0

        trans_values = [trans_dict.get(lc, {}).get(cat, 0) for lc in last_cats]
        trans_score = float(np.mean(trans_values)) if trans_values else 0

        scores[cat] = (
            0.28 * user_score +
            0.09 * demo_prob +
            0.07 * demo_lift +
            0.07 * province_prob +
            0.03 * province_lift +
            0.04 * dow_prob +
            0.04 * hour_prob +
            0.05 * brand_prob +
            0.04 * brand_lift +
            0.15 * pair_score +
            0.10 * trans_score +
            0.04 * global_score
        )

    return sorted(scores, key=scores.get, reverse=True)[:top_k]
```

```python
def evaluate_full(
    test_df,
    recommender_inputs,
    user_last_dict,
    user_last_cats,
    k=5
):
    (
        user_pref,
        segment_dict,
        province_dict,
        dow_dict,
        hour_dict,
        brand_segment_dict,
        global_dist,
        categories,
        pair_dict,
        trans_dict
    ) = recommender_inputs

    hits = 0
    total = 0
    mrr = 0
    ndcg = 0

    all_recs = []

    for _, row in test_df.iterrows():
        user_id = row['user_id']
        true_cat = row['category_name']

        user_row = user_last_dict.get(user_id)
        if user_row is None:
            continue

        recs = recommend_v2(
            user_id=user_id,
            user_row=user_row,
            user_pref=user_pref,
            segment_dict=segment_dict,
            global_dist=global_dist,
            categories=categories,
            pair_dict=pair_dict,
            trans_dict=trans_dict,
            user_last_cats=user_last_cats,
            province_dict=province_dict,
            dow_dict=dow_dict,
            hour_dict=hour_dict,
            brand_segment_dict=brand_segment_dict,
            top_k=k
        )

        all_recs.append(recs)

        if true_cat in recs:
            hits += 1
            rank = recs.index(true_cat) + 1
            mrr += 1 / rank
            ndcg += 1 / np.log2(rank + 1)

        total += 1

    hit_rate = hits / total
    mrr_score = mrr / total
    ndcg_score = ndcg / total

    recommended_items = {item for rec in all_recs for item in rec}
    coverage = len(recommended_items) / len(categories)

    novelty = np.mean([
        np.mean([-np.log2(global_dist.get(item, 1e-6)) for item in rec])
        for rec in all_recs
    ])

    return {
        f'HitRate@{k}': hit_rate,
        f'MRR@{k}': mrr_score,
        f'NDCG@{k}': ndcg_score,
        'Coverage': coverage,
        'Novelty': novelty
    }
```

```python
# ===== PREPROCESS =====
df = add_age_group(master_interaction)

# ===== SPLIT =====
train_df, test_df = temporal_split(df)

brand_segment_map = build_brand_segment_map(train_df)
train_df = apply_brand_segment(train_df, brand_segment_map)
test_df = apply_brand_segment(test_df, brand_segment_map)

# ===== BUILD BEHAVIORAL SIGNALS ON TRAIN ONLY =====
pair_stats_train = build_pair_stats(train_df) # tính lift giữa các cặp category dựa trên lịch sử mua hàng của người dùng trong tập train
transition_stats_train = build_transition_stats(train_df) # tính lift giữa các cặp category dựa trên thứ tự mua hàng của người dùng trong tập train

pair_dict = build_pair_dict(pair_stats_train) # xây dựng dictionary để lookup nhanh lift giữa các cặp category dựa trên lịch sử mua hàng của người dùng trong tập train
trans_dict = build_transition_dict(transition_stats_train) # xây dựng dictionary để lookup nhanh lift giữa các cặp category dựa trên thứ tự mua hàng của người dùng trong tập train

# ===== BUILD CONTEXT DISTRIBUTIONS ON TRAIN ONLY =====
# Tính phân phối category theo từng context (giới tính, độ tuổi, địa phương, thời gian, phân khúc thương hiệu) dựa trên tập train để tránh data leakage
global_dist, demo_dist, province_dist, dow_dist, hour_dist, brand_segment_dist = build_distributions(train_df)

segment_dict = build_context_dict(demo_dist, ['gender', 'age_group'])
province_dict = build_context_dict(province_dist, ['shipping_province_name'])
dow_dict = build_context_dict(dow_dist, ['day_of_week'])
hour_dict = build_context_dict(hour_dist, ['hour_bucket'])
brand_segment_dict = build_context_dict(brand_segment_dist, ['brand_segment'])

user_pref = build_user_pref(train_df)
categories = train_df['category_name'].unique()
user_last_cats = build_user_last_categories(train_df)

# ===== BUILD RECOMMENDER INPUTS =====
recommender_inputs = (
    user_pref,
    segment_dict,
    province_dict,
    dow_dict,
    hour_dict,
    brand_segment_dict,
    global_dist,
    categories,
    pair_dict,
    trans_dict
)

# ===== PRECOMPUTE USER LAST CONTEXT =====
user_last = (
    train_df.sort_values(['user_id', 'order_date', 'sale_order_id'])
    .groupby('user_id')
    .tail(1)
)
user_last_dict = user_last.set_index('user_id').to_dict('index')

# ===== EVALUATE =====
metrics = evaluate_full(
    test_df=test_df,
    recommender_inputs=recommender_inputs,
    user_last_dict=user_last_dict,
    user_last_cats=user_last_cats,
    k=5
)

metrics_df = pd.DataFrame({
    'metric': list(metrics.keys()),
    'value': list(metrics.values())
})
display(metrics_df)
print(metrics)
```
