## 2. Statistics

```python
import pandas as pd

files = [
    "export_csv/customer.csv",
    "export_csv/product.csv",
    "export_csv/product_category.csv",
    "export_csv/sale_order.csv",
    "export_csv/sale_order_line.csv",
]

dfs = {}

for file in files:
    dfs[file] = pd.read_csv(file)

for name, df in dfs.items():
    print(f"=== {name} ===")
    display(df.describe(include="all"))
```

## 3. Data Types

```python
for file, df in dfs.items():
    print(f"=== {file} ===")
    display(df.dtypes)
```

## 4. Find missing values

```python
import pandas as pd

for file in files:
    missing = pd.read_csv(file).isnull().sum()
    print(f"\n=== {file} ===")
    print(missing)
```

```python
# Check for duplicates in each DataFrame
for file, df in dfs.items():
    duplicates = df.duplicated().sum()
    print(f"\n=== {file} ===")
    print(f"Number of duplicate rows: {duplicates}")
```

```python
# remove nulls and duplicates
for file, df in dfs.items():
    df.dropna(inplace=True)
    df.drop_duplicates(inplace=True)
```

### Sale Order

```python
# draw boxplot and distribution plot for each numeric column in sale_order.csv
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sale_order_df = pd.read_csv('export_csv/sale_order.csv')    
numeric_cols = sale_order_df.select_dtypes(include=['float64', 'int64']).columns
numeric_cols = [col for col in numeric_cols if col != 'channel_id' and col != 'merchant_id' and col != 'status']  # exclude id columns
for col in numeric_cols:
    plt.figure(figsize=(12, 5))
    
    plt.subplot(1, 1, 1)
    sns.boxplot(x=sale_order_df[col])
    plt.title(f'Boxplot of {col}')
    
    plt.tight_layout()
    plt.show()
```

## 6. Bar chart

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import math

product_df = pd.read_csv("export_csv/product.csv")
product_category_df = pd.read_csv("export_csv/product_category.csv")

merged_df = product_df.merge(
    product_category_df[['category_id', 'category_name']],
    on='category_id',
    how='left'
)

category_counts = (
    merged_df.groupby(['category_id', 'category_name'])
    .size()
    .reset_index(name='product_count')
    .sort_values(by='category_id')
)

category_counts['category_name'] = category_counts['category_name'].fillna('Unknown')
category_counts['label'] = (
    category_counts['category_id'].astype(str) + ' - ' + category_counts['category_name']
)

n = len(category_counts)
chunk_size = math.ceil(n / 3)

chunks = [
    category_counts.iloc[i:i + chunk_size]
    for i in range(0, n, chunk_size)
]

fig, axes = plt.subplots(3, 1, figsize=(16, 18))

for i, chunk in enumerate(chunks):
    sns.barplot(data=chunk, x='label', y='product_count', ax=axes[i])
    axes[i].set_title(f'So luong san pham theo category - Phan {i+1}')
    axes[i].set_xlabel('Category')
    axes[i].set_ylabel('So luong san pham')
    axes[i].tick_params(axis='x', labelrotation=45)

plt.tight_layout()
plt.show()
```

### Top 10 category

```python
# top 10 category có nhiều sản phẩm nhất trong product.csv
import pandas as pd
product_df = pd.read_csv("export_csv/product.csv")
product_category_df = pd.read_csv("export_csv/product_category.csv")
merged_df = product_df.merge(
    product_category_df[['category_id', 'category_name']],
    on='category_id',
    how='left'
)
category_counts = (
    merged_df.groupby(['category_id', 'category_name'])
    .size()
    .reset_index(name='product_count')
    .sort_values(by='product_count', ascending=False)
    .head(10)
)
category_counts['category_name'] = category_counts['category_name'].fillna('Unknown')
category_counts['label'] = category_counts['category_name'] + ' (' + category_counts['product_count'].astype(str) + ')'
plt.figure(figsize=(12, 6))
sns.barplot(data=category_counts, x='label', y='product_count', palette='viridis')
plt.xlabel('Category')
plt.ylabel('Number of Products')
plt.title('Top 10 Categories with Most Products')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

### Top 10 Brands with Most Purchases

```python
# top 10 brand nhiều người mua nhất theo brand_name trong sale_order_line.csv, cần merge với product.csv và brand.csv để lấy brand_name
import pandas as pd
product_df = pd.read_csv("export_csv/product.csv")
brand_df = pd.read_csv("export_csv/brand.csv")
sale_order_line_df = pd.read_csv("export_csv/sale_order_line.csv")
merged_df = sale_order_line_df.merge(
    product_df[['product_id', 'brand_id']],
    on='product_id',
    how='left'
).merge(
    brand_df[['brand_id', 'brand_name']],
    on='brand_id',
    how='left'
)
brand_counts = (
    merged_df.groupby(['brand_id', 'brand_name'])
    .size()
    .reset_index(name='purchase_count')
    .sort_values(by='purchase_count', ascending=False)
    .head(10)
)
brand_counts['brand_name'] = brand_counts['brand_name'].fillna('Unknown')
brand_counts['label'] = brand_counts['brand_name'] + ' (' + brand_counts['purchase_count'].astype(str) + ')'
plt.figure(figsize=(12, 6))
sns.barplot(data=brand_counts, x='label', y='purchase_count', palette='viridis')
plt.xlabel('Brand')
plt.ylabel('Number of Purchases')
plt.title('Top 10 Brands with Most Purchases')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# số lần mua hàng trung bình của mỗi khách hàng trong sale_order.csv, cần merge với customer.csv để lấy thông tin khách hàng
import pandas as pd 
sale_order_df = pd.read_csv('export_csv/sale_order.csv')
customer_df = pd.read_csv('export_csv/customer.csv')
merged_df = sale_order_df.merge(
    customer_df[['user_id', 'age', 'gender']],
    on='user_id',   
    how='left'
)
purchase_counts = merged_df.groupby('user_id').size().reset_index(name='purchase_count')
average_purchases = purchase_counts['purchase_count'].mean()
print(f'So lan mua hang trung binh cua moi khach hang: {average_purchases:.2f}')
```

### Number of Orders by Month

```python
# phân bố theo thời gian tạo đơn hàng trong sale_order.csv, vẽ biểu đồ cột số lượng đơn hàng theo tháng
import pandas as pd
sale_order_df = pd.read_csv('export_csv/sale_order.csv')
sale_order_df['order_date'] = pd.to_datetime(sale_order_df['order_date'])
sale_order_df['month'] = sale_order_df['order_date'].dt.to_period('M')
monthly_order_counts = sale_order_df.groupby('month').size().reset_index(name='order_count')
plt.figure(figsize=(12, 6))
sns.barplot(x=monthly_order_counts['month'].astype(str), y=monthly_order_counts['order_count'], palette='viridis')
plt.xlabel('Month')
plt.ylabel('Number of Orders')
plt.title('Number of Orders by Month')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
